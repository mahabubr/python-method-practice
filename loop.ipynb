```python
text = "Hello Python 123"

vowels = 0
consonants = 0
digits = 0
spaces  = 0

for ch in text.lower():
    if ch in "aeiou":
        vowels += 1

    elif ch.isalpha():
        consonants += 1

    elif ch.isdigit():
        digits += 1

    elif ch == " ":
        spaces += 1

print(vowels, consonants, digits, spaces)
```

```python
text = "banana"

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1;
    else:
        freq[ch] = 1

freq
```

```python
students = [
    {"name": "Rahim", "marks": [80, 75, 90]},
    {"name": "Karim", "marks": [60, 85, 70]},
    {"name": "Hasan", "marks": [90, 95, 85]},
]

for student in students:
    name = student["name"]
    marks = student["marks"]

    total = 0

    for mark in marks:
        total += mark;

    avg = total / len(marks)

    print(f"Name: {name}, Marks: {avg:.2f}")
```

```python
number = 12345

total = 0

while number > 0:
    digit = number % 10
    total += digit
    number = number // 10

total
```

```python
matrix = [[10, 20, 30], [40, 50, 60], [70, 80, 90]]

for i in range(len(matrix)):
    total = 0

    for j in range(len(matrix[i])):
        total += matrix[i][j]

    print(total)
```

```python
students = {
    "Rahim": {"Math": 85, "English": 70, "Science": 90},
    "Karim": {"Math": 95, "English": 80, "Science": 75},
    "Hasan": {"Math": 75, "English": 95, "Science": 85},
}

highest = {}

for name, subjects in students.items():
    for subject, mark in subjects.items():
        if subject not in highest:
            highest[subject]= {"name": name, "mark": mark}
        elif mark > highest[subject]["mark"]:
            highest[subject] = {"name": name, "mark": mark}


highest
```

```python
numbers = [10, -5, 20, 0, -15, 30, 0, -8]

groups = {
    "positive": [],
    "negative": [],
    "zero": []
}

for num in numbers:
    if num > 0:
        groups["positive"].append(num);
    elif num < 0:
        groups["negative"].append(num);
    else:
        groups["zero"].append(num);

groups
```

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for num1 in list1:
    for num2 in list2:
        if num1 == num2:
            if num1 not in common:
                common.append(num1)

common
```
