```python
import matplotlib.pyplot as plt
```

```python
exams = ["Exam 1", "Exam 2", "Exam 3", "Exam 4", "Exam 5"]
marks = [65, 72, 80, 75, 90]

plt.plot(exams, marks)
plt.xlabel("Exams")
plt.ylabel("Marks")

plt.title("Exam Marks")

plt.grid(alpha=0.5)

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [12000, 15000, 10000, 18000, 22000]

plt.bar(months, sales)

plt.show()
```

```python
ages = [18, 19, 20, 21, 22, 20, 19, 23, 21, 22, 20, 19, 24, 21, 22, 23, 20, 18, 21, 22]

plt.hist(ages, bins=5)

plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [30, 32, 31, 29, 33, 35, 34]

plt.plot(days, temperature, marker="o")

plt.show()
```

```python
products = ["Mobile", "Laptop", "Tablet", "Headphone"]
sales = [40, 30, 20, 10]

plt.pie(sales, labels=products,)

plt.show()

plt.show()
```
