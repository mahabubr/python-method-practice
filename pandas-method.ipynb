```python
import pandas as pd
```

```python
data = {
    "name": ["Rahim", "Karim", "Sakib", "Nabil", "Hasan", "Rafi"],
    "department": ["CSE", "EEE", "CSE", "EEE", "CSE", "BBA"],
    "age": [22, 23, 24, 21, 25, 22],
    "salary": [30000, 35000, 40000, 32000, 45000, 28000],
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["salary"].agg([
    "sum", "mean", "max", "min"
])
```

```python
df.groupby(["department", "age"])["salary"].mean()
```

```python
df.groupby("department").agg({
    "salary": ["mean", "max"],
    "age": ["mean", 'min']
})
```

```python
df.groupby("department", as_index=False)["salary"].mean()
```

```python
df.groupby('department')["salary"].mean().reset_index()
```

```python
df['department_avg'] = df.groupby("department")["salary"].transform('mean')
```

```python
import pandas as pd

customers = pd.DataFrame(
    {
        "customer_id": [1, 2, 3, 4],
        "name": ["Rahim", "Karim", "Sakib", "Nabil"],
        "city": ["Dhaka", "Sylhet", "Dhaka", "Khulna"],
    }
)

orders = pd.DataFrame(
    {
        "order_id": [101, 102, 103, 104],
        "customer_id": [1, 2, 1, 5],
        "amount": [500, 800, 300, 900],
    }
)
```

```python
pd.merge(customers, orders, on="customer_id", how="inner")
```

```python
pd.merge(customers, orders, on="customer_id", how="left")
```

```python
pd.merge(customers, orders, on="customer_id", how="right")
```

```python
pd.merge(customers, orders, on="customer_id", how="outer")
```

```python
sales = pd.DataFrame({"buyer_id": [1, 2, 3], "total": [500, 700, 900]})
```

```python
pd.merge(customers, sales, left_on="customer_id", right_on="buyer_id", how="inner")
```

```python
df1 = pd.DataFrame(
    {"emp_id": [1, 2, 3], "dept_id": [10, 20, 10], "name": ["A", "B", "C"]}
)

df2 = pd.DataFrame(
    {"emp_id": [1, 2, 3], "dept_id": [10, 20, 20], "salary": [30000, 40000, 50000]}
)
```

```python
pd.merge(df1, df2, on=["emp_id", "dept_id"], how="inner")
```

```python
pd.merge(customers, customers, on="customer_id", suffixes=("_old", "_new"))
```

```python
import pandas as pd

df1 = pd.DataFrame(
    {"name": ["Rahim", "Karim"], "age": [22, 25], "salary": [30000, 40000]}
)

df2 = pd.DataFrame(
    {"name": ["Sakib", "Nabil"], "age": [24, 27], "salary": [35000, 45000]}
)
```

```python
pd.concat([df1, df2], axis = 0)
```

```python
pd.concat([df1, df2], ignore_index=True)
```

```python
personal = pd.DataFrame({"name": ["Rahim", "Karim"], "age": [22, 25]})

academic = pd.DataFrame({"marks": [85, 90], "grade": ["A", "A+"]})
```

```python
pd.concat([personal, academic], axis = 1)
```

```python
branch1 = pd.DataFrame({"product": ["Laptop", "Mobile"], "sales": [50000, 30000]})

branch2 = pd.DataFrame({"product": ["Tablet", "Watch"], "profit": [5000, 3000]})
```

```python
pd.concat([branch1, branch2], ignore_index=True, join='outer')
```

```python
pd.concat([branch1, branch2], join="inner", ignore_index=True)
```

```python
pd.concat([df1, df2], keys=["Branch_A", "Branch_B"])
```

```python
jan = df1
feb = df2
march = df1

pd.concat([jan, feb, march], ignore_index=True)
```

```python
import pandas as pd

df = pd.DataFrame(
    {
        "name": ["Rahim", "Karim", "Sakib", "Nabil"],
        "age": [22, 35, 17, 45],
        "salary": [30000, 50000, 20000, 80000],
        "marks": [85, 65, 90, 40],
    }
)
```

```python
import numpy as np
```

```python
df["marks"].apply(np.sqrt)
```

```python
df["salary"].apply(lambda x: x * 0.10)
```

```python
def age(age):
    if age < 18:
        return "Minor"
    elif age < 40:
        return "Adult"
    else:
        return "Senior"

df["category"] = df["age"].apply(age)

df
```

```python
df["total"] = df.apply(lambda row: row["marks"] + 10, axis  = 1)
```

```python
df
```

```python
df["name"] = df["name"].apply(lambda x: x.upper())
```
