```python
salaries = [20000, 30000, 40000, 50000]
```

```python
result = []

for salary in salaries:
    result.append(salary * 1.10)

result
```

```python
[salary * 1.10 for salary in salaries]
```

```python
list(map(lambda salary: salary * 1.10, salaries))
```

```python
def increase_salary(salary):
    return salary * 1.10

list(map(increase_salary, salaries))
```

```python
temperatures = [20, 25, 30, 35]

list(map(lambda c: (c * 9/5) + 32, temperatures))
```

```python
scores = [45, 60, 75, 90]

list(map(lambda sc: sc / 100, scores))
```
